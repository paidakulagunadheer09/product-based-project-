```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("earthquake_1995-2023.csv")

country_counts = df["country"].value_counts().head(10)

plt.figure(figsize=(10,5))

plt.bar(country_counts.index, country_counts.values)

plt.title("Top 10 Countries by Number of Earthquakes")
plt.xlabel("Country")
plt.ylabel("Number of Earthquakes")

plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("earthquake_1995-2023.csv")

country_counts = df["country"].value_counts().head(10)

plt.figure(figsize=(10, 5))

# Create purple-to-yellow colors using a colormap
colors = plt.cm.plasma(np.linspace(0, 1, 10))

plt.bar(country_counts.index, country_counts.values, color=colors)

plt.title("Top 10 Countries by Number of Earthquakes")
plt.xlabel("Country")
plt.ylabel("Number of Earthquakes")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# Load the dataset
df = pd.read_csv("earthquake_1995-2023.csv")

# Count earthquakes for each country
country_counts = df["country"].value_counts().head(10)

# Create normalization
norm = mpl.colors.Normalize(
    vmin=country_counts.min(),
    vmax=country_counts.max()
)

# Select colormap
cmap = mpl.cm.viridis

# Generate colors based on earthquake counts
bar_colors = cmap(norm(country_counts.values))

# Create bar chart
plt.figure(figsize=(10, 6))

plt.bar(
    country_counts.index,
    country_counts.values,
    color=bar_colors,
    edgecolor="black"
)

# Create ScalarMappable for colorbar
sm = mpl.cm.ScalarMappable(
    cmap=cmap,
    norm=norm
)
sm.set_array([])

# Add colorbar
plt.colorbar(
    sm,
    ax=plt.gca(),
    label="Number of Earthquakes"
)

# Add labels and title
plt.title("Top 10 Countries by Number of Earthquakes")
plt.xlabel("Country")
plt.ylabel("Number of Earthquakes")

plt.xticks(rotation=45)
plt.tight_layout()

# Display graph
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("/content/earthquake_1995-2023.csv")

# Create histogram
plt.figure(figsize=(10, 6))

plt.hist(
    df["magnitude"].dropna(),
    bins=20,
    color="purple",
    edgecolor="black",
    alpha=0.8
)

# Title and labels
plt.title(
    "Distribution of Earthquake Magnitudes",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Earthquake Magnitude", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

# Grid
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# Load the dataset
df = pd.read_csv("earthquake_1995-2023.csv")

# Count earthquakes for each country
country_counts = df["country"].value_counts().head(10)

# Create different shades of orange
cmap = mpl.cm.Oranges
colors = cmap([0.35, 0.42, 0.49, 0.56, 0.63,
               0.70, 0.77, 0.84, 0.91, 0.98])

# Plot donut chart
plt.figure(figsize=(10, 7))

plt.pie(
    country_counts.values,
    labels=country_counts.index,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4, edgecolor="white")
)

# Title
plt.title("Top 10 Countries by Number of Earthquakes")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("earthquake_1995-2023.csv")

# Remove missing values
data = df[["magnitude", "depth"]].dropna()

# Create scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(
    data["magnitude"],
    data["depth"],
    alpha=0.6
)

# Title and labels
plt.title("Earthquake Magnitude vs Depth")
plt.xlabel("Magnitude")
plt.ylabel("Depth")

# Grid
plt.grid(alpha=0.3)

plt.show()
```
